```python
import torch
from torch.nn.attention.flex_attention import flex_attention
```

```python
Q = torch.tensor([[[[1., 2, 3, 4], [1, 3, 4, 1]]]])
K = torch.tensor([[[[1., 1, 1, 1], [2, 2, 3, 1]]]])
V = torch.tensor([[[[1., 2, 1, 2], [3, 2, 1, 2]]]])
Q.shape, K.shape, V.shape
```

```python
torch.softmax(Q @ K.transpose(2, 3), dim=-1) @ V
```

```python
def score_op(score, d, h, q_idx, kv_idx):
    return score / 200

flex_attention(Q, K, V, score_mod=score_op)
```

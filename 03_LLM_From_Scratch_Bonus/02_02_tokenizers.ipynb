```python
import tiktoken
from transformers import AutoTokenizer
```

```python
tt_tokenizer = tiktoken.get_encoding("gpt2")
tt_tokenizer.n_vocab
```

```python
hf_tokenizer = AutoTokenizer.from_pretrained("openai-community/gpt2")
```

```python
txt = "Hello how are you Sir?"
hf_tokenizer(txt)["input_ids"]
```

```python
txt = "Hello how are you Sir?"
# %timeit tt_tokenizer.encode(txt)
tt_tokenizer.encode(txt)
```

```python
checkpoint = "HuggingFaceTB/SmolLM2-135M"
hf_tokenizer_smol = AutoTokenizer.from_pretrained(checkpoint)
```

```python
txt = "Hello how are you Sir?"
# %timeit hf_tokenizer_smol(txt)["input_ids"]
hf_tokenizer_smol(txt)["input_ids"]
```

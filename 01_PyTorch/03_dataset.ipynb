```python
import torch
from torch.utils.data import Dataset, DataLoader
```

```python
class ToyDataset(Dataset):

    def __init__(self, x: torch.Tensor, y: torch.Tensor):
        self.inputs = x
        self.labels = y

    def __getitem__(self, index):
        return self.inputs[0], self.labels[0]

    def __len__(self):
        return self.inputs.shape[0]
```

```python
X_train = torch.tensor(
    [
        [1.1, 1.3],
        [1.4, 1.9],
        [1.0, 1.9],
        [2.7, 0.3],
        [2.9, 0.3],
    ]
)
Y_train = torch.tensor([[1], [1], [1], [0], [0]])
X_train.shape, Y_train.shape
```

```python
X_test = torch.tensor([[1, 1], [2, 0]])
Y_test = torch.tensor([[1], [0]])
X_test.shape, Y_test.shape
```

```python
train_ds = ToyDataset(X_train, Y_train)
test_ds = ToyDataset(X_test, Y_test)
len(train_ds), len(test_ds)
```

```python
train_dl = DataLoader(train_ds, batch_size=2, shuffle=True, num_workers=0)
test_dl = DataLoader(test_ds, batch_size=2, shuffle=False, num_workers=0)
```

```python
for i, (x, y) in enumerate(train_dl):
    print(f"{i} | {x=} | {y=}")
```

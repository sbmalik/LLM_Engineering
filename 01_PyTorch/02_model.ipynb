```python
import torch
```

```python
class TestModel(torch.nn.Module):

    def __init__(self, ip_dims, out_dims):
        super().__init__()
        self.layers = torch.nn.Sequential(
            torch.nn.Linear(ip_dims, 30),
            torch.nn.ReLU(),
            torch.nn.Linear(30, 10),
            torch.nn.ReLU(),
            torch.nn.Linear(10, out_dims),
        )

    def forward(self, ips):
        return self.layers(ips)
```

```python
ips = torch.rand(1, 10)
ips.shape
```

```python
test_model = TestModel(10, 3)
test_model
```

```python
test_model(ips)
```

```python
trainable_param_count = 0
for p in test_model.parameters():
    if p.requires_grad == False:
        print(p)
    else:
        trainable_param_count += p.numel()

print(trainable_param_count)
```

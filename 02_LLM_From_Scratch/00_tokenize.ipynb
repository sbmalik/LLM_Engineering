```python
import urllib
import re
```

```python
# Download dataset


url = "https://raw.githubusercontent.com/rasbt/LLMs-from-scratch/main/ch02/01_main-chapter-code/the-verdict.txt"
file_path = "the-verdict.txt"
urllib.request.urlretrieve(url, file_path)
```

```python
# Check total character
with open("the-verdict.txt", "r") as file:
    data = file.read()

len(data.strip()), data[:10]
```

```python
# build a tokenization scheme
text = "Hello, world. Is this-- a test?"
result = re.split(r'([,.:;?_!"()\']|--|\s)', text)
# if you don't want extra space
result = [res.strip() for res in result if res.strip()]
result
```

```python
# tokenize real data
data_tokenized = re.split(r'([,.:;?_!"()\']|--|\s)', data)
data_tokenized = [datum.strip() for datum in data_tokenized if datum.strip()]
data_tokenized.extend(["<|endoftext|>", "<|unk|>"])
data_tokenized[-5::1]
```

```python
# convert this to token-ids (unique -> sort -> give ids)
data_tokenized = sorted(set(data_tokenized))
data_tk_ids_tti = {datum: idx for idx, datum in enumerate(data_tokenized)}
data_tk_ids_itt = {idx: datum for idx, datum in enumerate(data_tokenized)}
len(data_tk_ids_itt), data_tk_ids_tti, data_tk_ids_itt,
```

```python
# tokenizer class
class SimpleTokenizerV1:

    def __init__(self, vocab):
        self.str_to_int = vocab
        self.str_to_int_keys = list(self.str_to_int.keys())
        self.int_to_str = {i: s for s, i in vocab.items()}

    def encode(self, text):
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
        preprocessed = [datum.strip() for datum in preprocessed if datum.strip()]
        preprocessed = [
            item if item in self.str_to_int else "<|unk|>" for item in preprocessed
        ]
        return [self.str_to_int[s] for s in preprocessed]

    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])
        return re.sub(r'\s+([,.?!"()\'])', r"\1", text)
```

```python
tokenizer = SimpleTokenizerV1(data_tk_ids_tti)
tokenizer
```

```python
t1 = """
It's, how are you painting you know? Hello
"""
t2 = "Mister Jane wants to check something"
t3 = "<|endoftext|> ".join((t1, t2))
ids = tokenizer.encode(t3)
ids
```

```python
tokenizer.decode(ids)
```

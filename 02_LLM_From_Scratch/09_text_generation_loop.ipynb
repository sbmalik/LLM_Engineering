```python
import torch
import tiktoken
```

```python
tokenizer = tiktoken.get_encoding("gpt2")
```

```python
tokenizer.max_token_value, tokenizer.n_vocab
```

```python
def gpt_astra(inputs):
    return torch.randint(low=0, high=tokenizer.max_token_value, size=(1, 1))
```

```python
inputs = "Hello How are you Sir"
input_tokens = tokenizer.encode(inputs)
input_tokens
```

```python
msg = inputs
for _ in range(10):
    next_word = [gpt_astra(input_tokens).squeeze().item()]
    out_str = tokenizer.decode(next_word)
    input_tokens += [next_word]
    print(input_tokens)
    msg += out_str

msg
```
